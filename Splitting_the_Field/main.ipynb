```python
import pandas as pd
```

```python
df = pd.read_csv("baseball_positions.csv")
df.head()
```

```python
df['positions_list'] = df['Position'].str.split('/')

df_exploded = df.explode('positions_list')
df_exploded.reset_index(drop=True)
df_exploded.head()
```

```python
df_group = df_exploded.groupby('positions_list').agg({"positions_list": "count"})\
        .rename(columns={"positions_list": "positions_list_counter"})\
        .reset_index()
df_group = df_group.sort_values("positions_list_counter", ascending=False)
```
