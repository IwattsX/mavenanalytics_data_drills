```python
import pandas as pd
```

```python
df = pd.read_csv("marathon-data.csv")
df.head()
```

```python
df['final'] = pd.to_timedelta(df.final).dt.total_seconds() / 60.0
df.head()
```

```python
band_edges = [0] + [i for i in range(180, 390, 30)] + [float('inf')]
band_labels = ['sub 3', '3-3:30', '3:30-4', '4-4:30', '4:30-5', '5-5:30', '5:30-6', '6+']

df["band"] = pd.cut(
    df.final,
    bins=band_edges,
    labels=band_labels,
    right=False
)
df.head()
```

```python
df.band.value_counts(normalize=True)
```
