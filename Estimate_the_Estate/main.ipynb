```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("manhattan_property_sales.csv")
df.head()
```

```python
df['PRICE_PER_SQ_FOOT'] = df['SALE_PRICE'].replace(0, np.NaN) / df['SQUARE_FEET']
df.head()
```

```python
df['AVG_PRICE_PER_SQ_FOOT'] = df.groupby(['ZIP_CODE', 'BUILDING_CLASS'])['PRICE_PER_SQ_FOOT'].transform('mean')
df.head()
```

```python
df['MARKET_VALUE'] = np.where(df.SALE_PRICE > 0, df.SALE_PRICE, df.AVG_PRICE_PER_SQ_FOOT * df.SQUARE_FEET)
df.head()
```

```python
# Flag over 15 million in market value
len(df[df['MARKET_VALUE'] > 15_000_000])
```
