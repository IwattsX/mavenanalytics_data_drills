```python
import pandas as pd
from collections import defaultdict
from operator import attrgetter
```

```python
df = pd.read_csv("grocery_transactions.csv")
df.head()
```

```python
paired_df = pd.merge(df[['transaction_id', 'product_name']], df[['transaction_id', 'product_name']], on='transaction_id', how='outer')
paired_df = paired_df[paired_df['product_name_x'] != paired_df['product_name_y']]
paired_df.head()
```

```python
counter_hm = defaultdict(int) # counter dictionary for counting pairs
for product_name_x, product_name_y in zip(paired_df['product_name_x'], paired_df['product_name_y']):
    counter_hm[frozenset([product_name_x, product_name_y])] += 1

# correct products pairs that are counted twice
for k, v in counter_hm.items():
    counter_hm[k] = int(counter_hm[k]/2)
```

```python
class product_pairs:
    def __init__(self, s : frozenset, pairs_count):
        self.s = s
        self.pairs_count = pairs_count
    def __str__(self):
        return f"{self.s} pair has count: {self.pairs_count}"

pairs_list = []
for k, v in counter_hm.items():
    pairs_list.append(product_pairs(k, v))
result = sorted(pairs_list, key=attrgetter('pairs_count'), reverse=True)
for pair in result:
    print(pair)
    
```
