```python
import pandas as pd
```

```python
# price list histroy with pizz_id, effective_date, price
price_hist_df = pd.read_csv("price_history.csv", parse_dates=["effective_date"])
price_hist_df.head()
```

```python
# Get Products with pizza_id, name, current_price
products_df = pd.read_csv("products.csv")
products_df.head()
```

```python
# Transactions that have happened with order_detail_id. order_id, order_date, pizza_id, quantity
transactions_df = pd.read_csv("transactions.csv", parse_dates=["order_date"])
transactions_df.head()
```

```python
# merge transactions with the price_history
transactions_price_hist_df = pd.merge_asof(transactions_df,
    price_hist_df,
    left_on="order_date",
    right_on="effective_date",
    by="pizza_id",
    direction="backward"
)
transactions_price_hist_df.head()
```

```python
# get the sum of the transactions and round to get the flag
total = transactions_price_hist_df['price'] * transactions_price_hist_df['quantity']
total.sum()
```
